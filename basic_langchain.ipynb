```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)
```

```python
from pydantic import BaseModel, Field

class User(BaseModel):
    name: str = Field(description="User full name")
    age: int = Field(description="User age in years")
    email: str = Field(description="User email address")
```

```python
llm = llm.with_structured_output(User)
res=llm.invoke("hello, y Name Is Pradeep , age is 29 and email is psrana344@gmail.com . Can you please parse the the name, email and age")
```

```python
res.model_dump()
```

```python
from langchain.tools import tool

@tool
def send_email(email:str):
    """ I can send email for any use """
    return "Email Send Suceesfully"

@tool
def add_number(a:int, b:int):
    """ Add two number and return the sum """
    return a + b

@tool
def multiply_number(a:int, b:int):
    """ multply two number and return the product """
    return a * b



```

Agent

```python
from langchain.agents import create_agent
```

```python
agent = create_agent(
    model=llm,
    tools=[send_email, add_number, multiply_number],
    system_prompt="""
        You always call tools for user question.
    """
)
```

```python
agent
```

```python
que ="what is  2 * 3"
res = agent.invoke({"messages": [{"role":"user", "content":que}]})
```

```python
for msg in res["messages"]:
    msg.pretty_print()
```

```python
from langgraph.checkpoint.memory import InMemorySaver
agent = create_agent(
    model=llm,
    tools=[send_email, add_number, multiply_number],
    checkpointer = InMemorySaver(),
    system_prompt ="""
        You always call tools for user question.
    """
)
```

```python
while True:
    query = input("User: ")
    if query == "exit":
        break
    res = agent.invoke(
        {"messages": [{"role":"user", "content":query}]},
        {
            "configurable": {
                "thread_id": "user_1"
            }
        })

    ans = res["messages"][-1].content
    print("AI: ", ans)
```

Middleware

```python
from langchain.agents.middleware import SummarizationMiddleware
from langchain.agents import create_agent
CHAT = [
    "My order id is ORD-4471.",
    "It was a Nikon Z50 camera.",
    "The box came open.",
    "The lens cap is missing.",
    "I want a replacement, not a refund.",
    'I love react js, and want to learn it',
    'Who is PM of India ?'
]

agent = create_agent(
     
    system_prompt="""
        You are a helpfull assistant, reply in one short line
    """,
    model=llm,
    tools=[],
    checkpointer=InMemorySaver(),
    middleware=[
        SummarizationMiddleware(
            model=llm,
            trigger=("messages", 6),
            keep=("messages", 2)
        )
    ]
)
```

```python
config = {"configurable":{"thread_id":"user_2"}}
for msg in CHAT:
    
    agent.invoke(
        {"messages": [{"role":"user", "content":msg}]},
        config=config
    )
    print(f"Message in History:", len(agent.get_state(config).values['messages']))
```

```python
from langchain.agents.middleware import PIIMiddleware

agent = create_agent(
    model=llm,
    tools=[],
    middleware=[
        PIIMiddleware("email", strategy="redact", apply_to_input=True),
        PIIMiddleware("credit_card", strategy="redact", apply_to_input=True),
    ],
)
```

```python
query = "Hello, I received email from prateek@gmail.com for the payment of credit card 4111 1111 1111 1111"
res = agent.invoke({
    "messages": [ {"role":"user", "content":query} ]
})
```
